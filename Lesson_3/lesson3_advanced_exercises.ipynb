## Exercise 4: Identity vs equality

- Create two equal lists and compare with `==` and `is`.
- Assign one list to another variable and compare again.
- Write a short comment explaining differences in output.

```python
list1 = [1, 2, 3]
list2 = [1, 2, 3]

print("Comparing two distinct lists with identical elements:")
print(f"list1 == list2: {list1 == list2}")
print(f"list1 is list2: {list1 is list2}")

list3 = list1

print("\nComparing list1 and list3 (assigned reference):")
print(f"list1 == list3: {list1 == list3}")
print(f"list1 is list3: {list1 is list3}")

# Comment:
# `==` (checks whether the contents are equivalent).
# `is` (checks whether both variables point to the exact same memory address).
#
# - `list1 == list2` is True because both lists contain [1, 2, 3].
# - `list1 is list2` is False because they are two distinct objects stored in different memory locations.
# - `list1 is list3` is True because `list3 = list1` passes the reference directly, pointing to the exact same list in memory.
```

## Exercise 5: Membership filter

- Given `items = ["python", "java", "go", "rust"]`, test membership for 4 queries.
- Use both `in` and `not in`.
- Print human-readable results like `python: found`.

```python
items = ["python", "java", "go", "rust"]
queries = ["python", "ruby", "go", "swift"]

for query in queries:
    if query in items:
        print(f"{query}: found")
    if query not in items:
        print(f"{query}: not found")
```
